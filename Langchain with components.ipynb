```python
import os
os.environ["GROQ_API_KEY"] = "gsk_api key"
os.environ["TAVILY_API_KEY"] = "tvly-api key"

from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_groq import ChatGroq
from langchain_community.tools.tavily_search import TavilySearchResults
from langgraph.prebuilt import create_react_agent

# 1. LLM
llm = ChatGroq(model="openai/gpt-oss-20b", temperature=0)

# 2. Tools (NEW path)
search = TavilySearchResults(max_results=5)

# 3. Agent (NEW - replaces initialize_agent)
agent = create_react_agent(llm, [search])

# 4. SWOT Prompt (replaces LLMChain + RetrievalQA)
swot_prompt = PromptTemplate.from_template(
    "You are a Market Analyst. Company: {company}. Context: {context}. Give detailed SWOT."
)

# 5. Run
company = "Tesla"
context = search.invoke(f"{company} latest market news 2025")
chain = swot_prompt | llm | StrOutputParser()
result = chain.invoke({"company": company, "context": context})
print(result)
```
