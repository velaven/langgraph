```python
from langchain_core.tools import Tool
from tavily import TavilyClient

# Setup Tavily Client
tavily_client = TavilyClient(api_key=os.getenv("tvly-api-key"))

# Define a search function that Tavily Tool can use
def tavily_search(query: str) -> str:
    results = tavily_client.search(query=query, max_results=5)
    return "\n\n".join([r["content"] for r in results["results"]])

# Create LangChain Tool
tavily_tool = Tool(
    name="Tavily Web Search",
    func=tavily_search,
    description="Use this tool to search the web for fresh info about companies or markets"
)
```

```python
import os
os.environ["GROQ_API_KEY"] = "gsk_api-key"
os.environ["TAVILY_API_KEY"] = "tvly-api-key"

from langchain_groq import ChatGroq
from langchain_community.tools.tavily_search import TavilySearchResults
from langgraph.prebuilt import create_react_agent
from langchain_core.prompts import PromptTemplate

# 1. Your SWOT template
swot_prompt = PromptTemplate.from_template("""
You are a Market Analyst. Using the following search context, generate a detailed
SWOT analysis for {company}.

Search Context: {context}

Provide SWOT in clear bullet points.
""")

# 2. Modern LLM + Tool
llm = ChatGroq(model="openai/gpt-oss-20b", temperature=0)
search = TavilySearchResults(max_results=5)
agent = create_react_agent(llm, [search])

# 3. Run it
company = "Tesla"
# First get search context
search_result = search.invoke(f"{company} market position 2025")

# Then generate SWOT
final_swot = swot_prompt.format(company=company, context=search_result)
result = agent.invoke({"messages": [("human", final_swot)]})

print(result["messages"][-1].content)
```
